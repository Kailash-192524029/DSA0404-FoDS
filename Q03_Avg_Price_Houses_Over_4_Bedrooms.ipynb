# Q3 - Average Sale Price of Houses with More Than 4 Bedrooms (NumPy)

**Question:** Find the average sale price of houses with more than four bedrooms.

**Dataset:** `Exp9.csv` (columns `price`, `area`, `bedrooms`, ...).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose Exp9.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(file_name, encoding="utf-8-sig")
house_data = df[["bedrooms", "area", "price"]].to_numpy()   # columns: 0=bedrooms, 1=area, 2=price
print("house_data shape:", house_data.shape)
df.head()
```

## Step 2: Filter houses with more than 4 bedrooms and average their price

```python
bedrooms = house_data[:, 0]
price = house_data[:, 2]

mask = bedrooms > 4                     # boolean mask
print("Houses with more than 4 bedrooms:", mask.sum())
print("Average sale price (>4 bedrooms):", round(price[mask].mean(), 2))
print("Average sale price (all houses) :", round(price.mean(), 2))
```

```python
plt.figure(figsize=(6, 4))
plt.bar(["> 4 bedrooms", "All houses"], [price[mask].mean(), price.mean()], color=["tomato", "steelblue"])
plt.title("Average sale price")
plt.ylabel("Price")
plt.tight_layout(); plt.show()
```

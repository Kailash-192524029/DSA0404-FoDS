# Q14 - Frequency Distribution of Customer Ages

**Question:** Find the frequency distribution of the ages of customers who made a purchase in the past month.

**Dataset:** `20_AB_Testing_Website_Dataset.csv` (`Age`, `Converted` = 1 means the visitor purchased, `Visit_Date`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose 20_AB_Testing_Website_Dataset.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(file_name)
df["Visit_Date"] = pd.to_datetime(df["Visit_Date"])
print(df.shape)
df.head()
```

## Step 2: Customers who made a purchase in the past month
'Past month' = the last 30 days in the dataset.

```python
last_date = df["Visit_Date"].max()
start_date = last_date - pd.Timedelta(days=30)
buyers = df[(df["Converted"] == 1) & (df["Visit_Date"] > start_date)]
print("Period:", start_date.date(), "to", last_date.date())
print("Purchasing customers:", len(buyers))
```

## Step 3: Frequency of each age

```python
age_freq = buyers["Age"].value_counts().sort_index()
print(age_freq.to_string())

plt.figure(figsize=(12, 5))
plt.bar(age_freq.index, age_freq.values, color="steelblue")
plt.title("Frequency distribution of customer ages (past month)")
plt.xlabel("Age"); plt.ylabel("Number of customers")
plt.show()
```

## Step 4: Age groups

```python
bins = [17, 25, 35, 45, 55, 65]
labels = ["18-25", "26-35", "36-45", "46-55", "56-65"]
groups = pd.cut(buyers["Age"], bins=bins, labels=labels).value_counts().sort_index()
print(groups)

groups.plot(kind="bar", color="seagreen", figsize=(7, 4), rot=0)
plt.title("Customers by age group (past month)")
plt.xlabel("Age group"); plt.ylabel("Number of customers")
plt.show()
```

# Q1 - Average Score per Subject (NumPy)

**Question:** Calculate the average score for each subject and find the subject with the highest average, using NumPy arrays.

**Dataset:** `student_scores.csv` (columns Math, Science, English, History).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose student_scores.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(file_name, encoding="utf-8-sig")
subjects = ["Math", "Science", "English", "History"]
student_scores = df[subjects].to_numpy()          # rows = students, columns = subjects
print("Shape of student_scores:", student_scores.shape)
df.head()
```

## Step 2: Average score for each subject

```python
avg_scores = student_scores.mean(axis=0)          # axis=0 -> average down each column
for sub, avg in zip(subjects, avg_scores):
    print(f"{sub:8s}: {avg:.2f}")
```

## Step 3: Subject with the highest average

```python
best = np.argmax(avg_scores)
print("Highest average subject:", subjects[best], "->", round(avg_scores[best], 2))
```

```python
plt.figure(figsize=(7, 4))
plt.bar(subjects, avg_scores, color="steelblue")
plt.title("Average score per subject")
plt.xlabel("Subject"); plt.ylabel("Average score")
plt.tight_layout(); plt.show()
```

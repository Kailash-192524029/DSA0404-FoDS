# Q7 - Order Data Analysis (Pandas)

**Question:** Find (1) the total number of orders by each customer, (2) the average order quantity for each product, (3) the earliest and latest order dates.

**Dataset:** `Exp7.csv` (columns `customer_name`, `order_id`, `order_date`, `product_name`, `quantity`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose Exp7.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# utf-8-sig removes the hidden BOM character at the start of the file
order_data = pd.read_csv(file_name, encoding="utf-8-sig")
# the file mixes dd/mm/yyyy and dd-mm-yyyy, so let pandas infer each date (day first)
order_data["order_date"] = pd.to_datetime(order_data["order_date"], format="mixed", dayfirst=True)
print("Rows:", len(order_data))
order_data[["customer_name", "order_date", "product_name", "quantity"]].head()
```

## 1. Total number of orders by each customer

```python
orders_per_customer = order_data.groupby("customer_name")["order_id"].nunique().sort_values(ascending=False)
print(orders_per_customer.head(10))
```

## 2. Average order quantity for each product

```python
avg_qty = order_data.groupby("product_name")["quantity"].mean().sort_values(ascending=False)
print(avg_qty.head(10).round(2))
```

## 3. Earliest and latest order dates

```python
print("Earliest order date:", order_data["order_date"].min().date())
print("Latest order date  :", order_data["order_date"].max().date())
```

```python
orders_per_customer.head(10).plot(kind="bar", figsize=(10, 5), color="steelblue")
plt.title("Top 10 customers by number of orders")
plt.xlabel("Customer"); plt.ylabel("Orders")
plt.xticks(rotation=45, ha="right")
plt.tight_layout(); plt.show()
```

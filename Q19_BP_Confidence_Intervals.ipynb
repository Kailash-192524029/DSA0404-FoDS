# Q19 - 95% Confidence Interval for Blood Pressure Reduction

**Question:** Find the 95% confidence interval for the mean reduction in blood pressure for the Drug group and for the Placebo group.

**Dataset:** `19_blood_pressure_trial_500_patients.csv` (columns `Group` and `Reduction`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose 19_blood_pressure_trial_500_patients.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

df = pd.read_csv(file_name)
print(df.shape)
print(df["Group"].value_counts())
df.head()
```

## Step 2: 95% confidence interval for each group (t-distribution)

```python
confidence = 0.95
results = {}

for group in ["Drug", "Placebo"]:
    x = df[df["Group"] == group]["Reduction"]
    n = len(x)
    mean = x.mean()
    sem = stats.sem(x)                                   # standard error
    low, high = stats.t.interval(confidence, df=n - 1, loc=mean, scale=sem)
    results[group] = (mean, low, high)
    print(f"{group}: n = {n}, mean reduction = {mean:.2f} mmHg")
    print(f"   95% CI = ({low:.2f}, {high:.2f})\n")
```

## Step 3: Plot the intervals

```python
groups = list(results.keys())
means = [results[g][0] for g in groups]
errors = [results[g][0] - results[g][1] for g in groups]

plt.figure(figsize=(6, 5))
plt.bar(groups, means, yerr=errors, capsize=10, color=["seagreen", "lightcoral"])
plt.title("Mean BP reduction with 95% confidence interval")
plt.ylabel("Reduction (mmHg)")
plt.show()
```

## Step 4: Conclusion

```python
d_low, d_high = results["Drug"][1], results["Drug"][2]
p_low, p_high = results["Placebo"][1], results["Placebo"][2]
if d_low > p_high or p_low > d_high:
    print("The two intervals do NOT overlap, so the drug and placebo groups differ clearly.")
else:
    print("The two intervals overlap, so the difference is not clear from the intervals alone.")
```

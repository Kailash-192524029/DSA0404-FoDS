# Question 11 – Monthly Sales Visualization Using Matplotlib

**Question:**
1. Write code to create a simple line plot to visualize sales over a month.
2. Write code to create a scatter plot to visualize sales over a month.
3. Develop a Python program to create a bar plot of monthly sales data.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
monthly_sales = [12000, 15000, 13500, 18000, 20000, 17500,
                 22000, 24000, 21000, 26000, 28000, 30000]

# 1. Line plot
plt.figure(figsize=(8, 4))
plt.plot(months, monthly_sales, marker="o")
plt.title("Monthly Sales - Line Plot")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.tight_layout()
plt.show()

# 2. Scatter plot
plt.figure(figsize=(8, 4))
plt.scatter(months, monthly_sales)
plt.title("Monthly Sales - Scatter Plot")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.tight_layout()
plt.show()

# 3. Bar plot
plt.figure(figsize=(8, 4))
plt.bar(months, monthly_sales)
plt.title("Monthly Sales - Bar Plot")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

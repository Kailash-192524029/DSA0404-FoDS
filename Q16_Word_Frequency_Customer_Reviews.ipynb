# Q16 - Word Frequency in Customer Reviews

**Question:** Calculate the frequency distribution of words in the customer reviews dataset.

**Dataset:** `17customer_reviews_FINAL.csv` (column `Review Text`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose 17customer_reviews_FINAL.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import re
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

df = pd.read_csv(file_name, encoding="latin-1")
reviews = df["Review Text"].dropna().astype(str)      # skip empty reviews
print("Number of reviews:", len(reviews))
reviews.head()
```

## Step 2: Clean the text and count words

```python
text = " ".join(reviews).lower()
text = re.sub(r"[^a-z\s]", " ", text)
words = text.split()

freq = Counter(words)
print("Total words :", len(words))
print("Unique words:", len(freq))

top = pd.DataFrame(freq.most_common(20), columns=["Word", "Frequency"])
print(top)
```

```python
plt.figure(figsize=(10, 5))
plt.bar(top["Word"], top["Frequency"], color="steelblue")
plt.xticks(rotation=45)
plt.title("Top 20 words in customer reviews")
plt.xlabel("Word"); plt.ylabel("Frequency")
plt.tight_layout(); plt.show()
```

## Step 3: Without stop words (more meaningful words)

```python
clean_words = [w for w in words if w not in ENGLISH_STOP_WORDS and len(w) > 2]
top2 = pd.DataFrame(Counter(clean_words).most_common(20), columns=["Word", "Frequency"])
print(top2)

plt.figure(figsize=(10, 5))
plt.bar(top2["Word"], top2["Frequency"], color="seagreen")
plt.xticks(rotation=45)
plt.title("Top 20 words (stop words removed)")
plt.xlabel("Word"); plt.ylabel("Frequency")
plt.tight_layout(); plt.show()
```

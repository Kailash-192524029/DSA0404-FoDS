```python
import numpy as np

student_scores = np.array([
    [85, 90, 78, 88],
    [92, 85, 80, 90],
    [78, 88, 85, 82],
    [90, 92, 88, 95]
])

subjects = ["Math", "Science", "English", "History"]

avg = np.mean(student_scores, axis=0)

print("Average marks:")
for i in range(4):
    print(subjects[i], ":", avg[i])

highest = np.argmax(avg)
print("Highest average subject:", subjects[highest])
```

```python
import numpy as np

# Columns: Bedrooms, Area, Sale Price
house_data = np.array([
    [3, 1200, 250000],
    [5, 2000, 450000],
    [4, 1500, 300000],
    [6, 2500, 600000],
    [5, 2200, 500000]
])

# Select houses with more than 4 bedrooms
selected = house_data[house_data[:, 0] > 4]

# Calculate average sale price
avg_price = np.mean(selected[:, 2])

print("Average sale price:", avg_price)
```

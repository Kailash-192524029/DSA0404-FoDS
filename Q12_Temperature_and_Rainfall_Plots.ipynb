# Q12 - Temperature and Rainfall Plots (Matplotlib)

**Question:** (1) Create a line plot of the monthly temperature data. (2) Create a scatter plot of the monthly rainfall data.

**Dataset:** none was provided, so a 12-month sample dataset is typed in below. Replace the values (or load your own CSV with `pd.read_csv`) to use real data.

Run the cells from top to bottom.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Temperature": [25.5, 26.8, 28.9, 31.2, 33.4, 33.0, 31.5, 31.0, 30.4, 28.6, 26.5, 25.2],   # degrees C
    "Rainfall":    [24, 8, 12, 15, 52, 53, 83, 124, 118, 267, 407, 139],                       # mm
})
data
```

## 1. Line plot of monthly temperature

```python
plt.figure(figsize=(9, 5))
plt.plot(data["Month"], data["Temperature"], marker="o", color="orangered")
plt.title("Monthly temperature (line plot)")
plt.xlabel("Month"); plt.ylabel("Temperature (°C)")
plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

## 2. Scatter plot of monthly rainfall

```python
plt.figure(figsize=(9, 5))
plt.scatter(data["Month"], data["Rainfall"], color="royalblue", s=70)
plt.title("Monthly rainfall (scatter plot)")
plt.xlabel("Month"); plt.ylabel("Rainfall (mm)")
plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

# Q12 - Temperature and Rainfall Plots (Matplotlib)

**Question:** (1) Create a line plot of the monthly temperature data. (2) Create a scatter plot of the monthly rainfall data.

**Dataset:** `weather_data.csv` (columns `Month`, `Temperature`, `Rainfall`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Load the dataset (weather_data.csv)
# If it is already in this Colab session it is reused; otherwise a file picker appears - choose weather_data.csv
import glob
found = glob.glob("/content/*weather_data*.csv")
if found:
    file_name = found[0]
else:
    from google.colab import files
    uploaded = files.upload()
    file_name = next(iter(uploaded))
print("Using file:", file_name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv(file_name, encoding="utf-8-sig")
data
```

## 1. Line plot of monthly temperature

```python
plt.figure(figsize=(9, 5))
plt.plot(data["Month"], data["Temperature"], marker="o", color="orangered")
plt.title("Monthly temperature (line plot)")
plt.xlabel("Month"); plt.ylabel("Temperature (°C)")
plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

## 2. Scatter plot of monthly rainfall

```python
plt.figure(figsize=(9, 5))
plt.scatter(data["Month"], data["Rainfall"], color="royalblue", s=70)
plt.title("Monthly rainfall (scatter plot)")
plt.xlabel("Month"); plt.ylabel("Rainfall (mm)")
plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

# Q5 - Average Fuel Efficiency and % Improvement (NumPy)

**Question:** Calculate the average fuel efficiency and the percentage improvement in fuel efficiency between two car models.

**Dataset:** `Fuel_Efficiency_datasets.csv` (column `highway_mpg(Fuel Efficiency)`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose Fuel_Efficiency_datasets.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(file_name, encoding="utf-8-sig")
df = df.dropna(subset=["highway_mpg(Fuel Efficiency)"]).reset_index(drop=True)
fuel_efficiency = df["highway_mpg(Fuel Efficiency)"].to_numpy()
print("Number of car models:", len(fuel_efficiency))
df[["make", "model", "highway_mpg(Fuel Efficiency)"]].head()
```

## Step 2: Average fuel efficiency

```python
print("Average fuel efficiency (mpg):", round(fuel_efficiency.mean(), 2))
```

## Step 3: Percentage improvement between two car models
Change `idx_a` and `idx_b` to compare any two models (row numbers in the table above).

```python
idx_a, idx_b = 5, 0                                   # model A (old) and model B (new)
a, b = fuel_efficiency[idx_a], fuel_efficiency[idx_b]

print("Model A:", df.loc[idx_a, "make"], df.loc[idx_a, "model"], "->", a, "mpg")
print("Model B:", df.loc[idx_b, "make"], df.loc[idx_b, "model"], "->", b, "mpg")
print("Improvement of B over A:", round((b - a) / a * 100, 2), "%")
```

```python
labels = [f"{df.loc[idx_a,'make']} {df.loc[idx_a,'model']}", f"{df.loc[idx_b,'make']} {df.loc[idx_b,'model']}"]
plt.figure(figsize=(6, 4))
plt.bar(labels, [a, b], color=["gray", "seagreen"])
plt.axhline(fuel_efficiency.mean(), color="red", linestyle="--", label="Average")
plt.title("Fuel efficiency of two models"); plt.ylabel("mpg"); plt.legend()
plt.tight_layout(); plt.show()
```

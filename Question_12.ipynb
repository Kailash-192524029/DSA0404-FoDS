# Question 12 – Monthly Temperature and Rainfall Visualization

**Question:**
1. Develop a Python program to create a line plot of monthly temperature data.
2. Develop a Python program to create a scatter plot of monthly rainfall data.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
temperature = [22, 24, 28, 32, 35, 34, 31, 30, 29, 27, 24, 22]  # °C
rainfall = [20, 15, 25, 45, 80, 150, 180, 170, 140, 100, 55, 25]  # mm

# 1. Line plot of monthly temperature
plt.figure(figsize=(8, 4))
plt.plot(months, temperature, marker="o")
plt.title("Monthly Temperature")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.tight_layout()
plt.show()

# 2. Scatter plot of monthly rainfall
plt.figure(figsize=(8, 4))
plt.scatter(months, rainfall)
plt.title("Monthly Rainfall")
plt.xlabel("Month")
plt.ylabel("Rainfall (mm)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

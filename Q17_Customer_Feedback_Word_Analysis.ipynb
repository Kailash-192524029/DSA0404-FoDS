# Q17 - Customer Feedback Word Frequency Program

**Question:** Load feedback data, remove punctuation, lowercase, remove stop words, find word frequencies, show top N words (N from user input) and plot a bar graph.

**Dataset:** `17customer_reviews_FINAL.csv` - the `Review Text` column is used as the `feedback` column.

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose 17customer_reviews_FINAL.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

## Step 2: Load the data (single column named `feedback`)

```python
import re
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

df = pd.read_csv(file_name, encoding="latin-1")
df = df[["Review Text"]].rename(columns={"Review Text": "feedback"}).dropna()
print("Feedback rows:", len(df))
df.head()
```

## Step 3: Preprocess (lowercase, remove punctuation, remove stop words)

```python
def preprocess(text):
    text = text.lower()                              # lowercase
    text = re.sub(r"[^a-z\s]", " ", text)           # remove punctuation/numbers
    return [w for w in text.split()
            if w not in ENGLISH_STOP_WORDS and len(w) > 1]   # remove stop words

all_words = []
for comment in df["feedback"]:
    all_words.extend(preprocess(comment))

freq = Counter(all_words)
print("Words after cleaning:", len(all_words))
print("Unique words:", len(freq))
```

## Step 4: Ask the user for N, show and plot the top N words

```python
try:
    N = int(input("Enter N (how many top words to show): "))
except ValueError:
    N = 10
    print("Invalid input, using N = 10")

top_n = freq.most_common(N)
print(f"\nTop {N} words:")
for rank, (word, count) in enumerate(top_n, start=1):
    print(f"{rank:>2}. {word:<15} {count}")

words_list = [w for w, c in top_n]
counts_list = [c for w, c in top_n]

plt.figure(figsize=(10, 5))
plt.bar(words_list, counts_list, color="steelblue")
plt.xticks(rotation=45)
plt.title(f"Top {N} most frequent words")
plt.xlabel("Word"); plt.ylabel("Frequency")
plt.tight_layout(); plt.show()
```

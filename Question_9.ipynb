# Question 9 – Analyze Property Data Using Pandas

**Question:** Using a DataFrame named property_data, find (1) the average listing price in each location, (2) the number of properties with more than four bedrooms, and (3) the property with the largest area.

```python
import pandas as pd

# Sample property data
property_data = pd.DataFrame({
    "property_id": [1, 2, 3, 4, 5, 6],
    "location": ["Chennai", "Tambaram", "Chennai", "Velachery",
                 "Tambaram", "Velachery"],
    "bedrooms": [3, 5, 4, 6, 2, 5],
    "area_sqft": [1200, 2400, 1800, 3000, 950, 2700],
    "listing_price": [6500000, 9000000, 7800000, 12000000, 4500000, 10500000]
})

# 1. Average listing price in each location
print("Average listing price by location:")
print(property_data.groupby("location")["listing_price"].mean())

# 2. Number of properties with more than four bedrooms
count = (property_data["bedrooms"] > 4).sum()
print("\nProperties with more than four bedrooms:", count)

# 3. Property with the largest area
largest_property = property_data.loc[property_data["area_sqft"].idxmax()]
print("\nProperty with the largest area:")
print(largest_property)
```

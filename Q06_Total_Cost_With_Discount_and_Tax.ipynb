# Q6 - Total Cost with Discount and Tax

**Question:** Use arithmetic operations to calculate the total cost of a customer's purchase, including discounts and taxes, given item prices, quantities, discount rate and tax rate.

**Dataset:** `grocery_items.csv` (columns `Item`, `Price`, `Quantity`). Discount and tax rates are set in the code.

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Load the dataset (grocery_items.csv)
# If it is already in this Colab session it is reused; otherwise a file picker appears - choose grocery_items.csv
import glob
found = glob.glob("/content/*grocery_items*.csv")
if found:
    file_name = found[0]
else:
    from google.colab import files
    uploaded = files.upload()
    file_name = next(iter(uploaded))
print("Using file:", file_name)
```

```python
import pandas as pd

df = pd.read_csv(file_name, encoding="utf-8-sig")
items      = df["Item"].tolist()
prices     = df["Price"].tolist()          # price per unit
quantities = df["Quantity"].tolist()

discount_rate = 10                         # percent
tax_rate      = 5                          # percent
df
```

## Step 2: Calculate the bill

```python
line_totals = [p * q for p, q in zip(prices, quantities)]
subtotal = sum(line_totals)
discount = subtotal * discount_rate / 100
after_discount = subtotal - discount
tax = after_discount * tax_rate / 100            # tax is charged on the discounted amount
total_cost = after_discount + tax

bill = pd.DataFrame({"Item": items, "Price": prices, "Qty": quantities, "Line Total": line_totals})
print(bill)
print()
print(f"Subtotal        : {subtotal:.2f}")
print(f"Discount ({discount_rate}%) : -{discount:.2f}")
print(f"After discount  : {after_discount:.2f}")
print(f"Tax ({tax_rate}%)       : +{tax:.2f}")
print(f"TOTAL COST      : {total_cost:.2f}")
```

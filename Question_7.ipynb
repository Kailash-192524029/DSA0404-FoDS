# Question 7 – Analyze Customer Orders Using Pandas

**Question:** Using a Pandas DataFrame named order_data, find (1) the total number of orders made by each customer, (2) the average order quantity for each product, and (3) the earliest and latest order dates.

```python
import pandas as pd

# Sample order data
order_data = pd.DataFrame({
    "customer_id": [101, 102, 101, 103, 102, 101],
    "order_date": ["2026-01-05", "2026-01-07", "2026-01-10",
                   "2026-01-12", "2026-01-15", "2026-01-20"],
    "product_name": ["Pen", "Book", "Book", "Pencil", "Pen", "Pen"],
    "order_quantity": [2, 1, 3, 5, 4, 1]
})

order_data["order_date"] = pd.to_datetime(order_data["order_date"])

# 1. Total number of orders made by each customer
print("Total orders by each customer:")
print(order_data.groupby("customer_id")["order_date"].count())

# 2. Average order quantity for each product
print("\nAverage order quantity for each product:")
print(order_data.groupby("product_name")["order_quantity"].mean())

# 3. Earliest and latest order dates
print("\nEarliest order date:", order_data["order_date"].min().date())
print("Latest order date:", order_data["order_date"].max().date())
```

# Q10 - Line Plot and Bar Plot of Monthly Sales (Matplotlib)

**Question:** (1) Create a line plot of the monthly sales data. (2) Create a bar plot of the monthly sales data.

**Dataset:** `Exp10.csv` (columns `order_date`, `sales`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose Exp10.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(file_name, encoding="utf-8-sig")
df["order_date"] = pd.to_datetime(df["order_date"], format="mixed", dayfirst=True)
df["sales"] = df["sales"].astype(str).str.replace(",", "").astype(float)   # values like "1,648" are text in the CSV

# Monthly sales for one year (change the year if you like)
year = 2014
monthly = (df[df["order_date"].dt.year == year]
           .groupby(df["order_date"].dt.month)["sales"].sum())
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
print(pd.DataFrame({"Month": months, "Sales": monthly.values.round(0)}))
```

## 1. Line plot

```python
plt.figure(figsize=(9, 5))
plt.plot(months, monthly.values, marker="o", color="steelblue")
plt.title(f"Monthly sales - {year} (line plot)")
plt.xlabel("Month"); plt.ylabel("Sales")
plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

## 2. Bar plot

```python
plt.figure(figsize=(9, 5))
plt.bar(months, monthly.values, color="seagreen")
plt.title(f"Monthly sales - {year} (bar plot)")
plt.xlabel("Month"); plt.ylabel("Sales")
plt.tight_layout(); plt.show()
```

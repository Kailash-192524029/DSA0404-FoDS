# Q2 - Average Price of Products Sold (NumPy)

**Question:** Find the average price of all the products sold in the past month using NumPy.

**Dataset:** `Sales.xlsx` (columns `Product_Name`, `Price`, `Month_sales`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose Sales.xlsx)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(file_name)
df["Month_sales"] = pd.to_datetime(df["Month_sales"])

# "Past month" = the most recent month that has full data in the file (Dec 2023)
last_month = df[(df["Month_sales"].dt.year == 2023) & (df["Month_sales"].dt.month == 12)]
print("Rows in the month:", len(last_month))
last_month.head()
```

## Step 2: Average price of all products sold

```python
prices = last_month["Price"].to_numpy()
print("Average price of all products sold:", round(prices.mean(), 2))
```

## Step 3: Average price per product (one row per product, like the 3x3 matrix idea)

```python
names = last_month["Product_Name"].unique()
avg_by_product = np.array([prices[(last_month["Product_Name"] == n).to_numpy()].mean() for n in names])
print(pd.DataFrame({"Product": names, "Average Price": avg_by_product.round(2)}))
```

```python
plt.figure(figsize=(7, 4))
plt.bar(names, avg_by_product, color="seagreen")
plt.title("Average price per product (past month)")
plt.xlabel("Product"); plt.ylabel("Average price")
plt.tight_layout(); plt.show()
```

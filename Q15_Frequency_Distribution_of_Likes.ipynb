# Q15 - Frequency Distribution of Likes Among Posts

**Question:** Calculate the frequency distribution of likes among the posts.

**Dataset:** `17customer_reviews_FINAL.csv` - each review is treated as a 'post' and `Positive Feedback Count` is treated as its number of likes (no real 'likes' column was provided).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose 17customer_reviews_FINAL.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(file_name, encoding="latin-1")
likes = df["Positive Feedback Count"]          # likes per post
print(df.shape)
df[["Clothing ID", "Positive Feedback Count"]].head()
```

## Step 2: Frequency of each like count

```python
like_freq = likes.value_counts().sort_index()
table = like_freq.reset_index()
table.columns = ["Likes", "Number of posts"]
print(table.to_string(index=False))

plt.figure(figsize=(10, 5))
plt.bar(like_freq.index, like_freq.values, color="steelblue")
plt.title("Frequency distribution of likes")
plt.xlabel("Likes on a post"); plt.ylabel("Number of posts")
plt.show()
```

## Step 3: Likes in ranges

```python
bins = [-1, 0, 2, 5, 10, likes.max()]
labels = ["0", "1-2", "3-5", "6-10", "10+"]
ranges = pd.cut(likes, bins=bins, labels=labels).value_counts().sort_index()
print(ranges)

ranges.plot(kind="bar", color="seagreen", figsize=(7, 4), rot=0)
plt.title("Posts grouped by number of likes")
plt.xlabel("Likes"); plt.ylabel("Number of posts")
plt.show()
```

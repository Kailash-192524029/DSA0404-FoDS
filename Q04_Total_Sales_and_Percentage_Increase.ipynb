# Q4 - Yearly Total Sales and % Increase Q1 to Q4 (NumPy)

**Question:** Calculate the total sales for the year and the percentage increase in sales from the first quarter to the fourth quarter.

**Dataset:** `Sales.xlsx` (columns `Month_sales`, `Total`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose Sales.xlsx)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(file_name)
df["Month_sales"] = pd.to_datetime(df["Month_sales"])
df = df[df["Month_sales"].dt.year == 2023]            # one full year (the file has a single stray row in 2024)

quarterly = df.groupby(df["Month_sales"].dt.quarter)["Total"].sum()
sales_data = quarterly.to_numpy()                      # one element per quarter
print(pd.DataFrame({"Quarter": ["Q1", "Q2", "Q3", "Q4"], "Sales": sales_data.round(2)}))
```

## Step 2: Total sales and percentage increase

```python
total_sales = sales_data.sum()
pct_increase = (sales_data[3] - sales_data[0]) / sales_data[0] * 100

print("Total sales for the year:", round(total_sales, 2))
print("Percentage increase Q1 -> Q4:", round(pct_increase, 2), "%")
```

```python
plt.figure(figsize=(7, 4))
plt.bar(["Q1", "Q2", "Q3", "Q4"], sales_data, color="steelblue")
plt.title("Quarterly sales (2023)")
plt.xlabel("Quarter"); plt.ylabel("Sales")
plt.tight_layout(); plt.show()
```

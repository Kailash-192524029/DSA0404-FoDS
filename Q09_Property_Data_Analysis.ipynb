# Q9 - Property Data Analysis (Pandas)

**Question:** Find (1) average listing price by location, (2) number of properties with more than four bedrooms, (3) the property with the largest area.

**Dataset:** `Exp9.csv` (columns `price`, `area`, `bedrooms`, `prefarea`, ...). The file has no `location` column, so `prefarea` (preferred area: yes/no) is used as the location group.

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose Exp9.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

property_data = pd.read_csv(file_name, encoding="utf-8-sig")
print("Properties:", len(property_data))
property_data.head()
```

## 1. Average listing price by location

```python
avg_price = property_data.groupby("prefarea")["price"].mean().round(2)
print(avg_price)
```

## 2. Properties with more than four bedrooms

```python
count_4plus = (property_data["bedrooms"] > 4).sum()
print("Properties with more than 4 bedrooms:", count_4plus)
```

## 3. Property with the largest area

```python
largest = property_data.loc[property_data["area"].idxmax()]
print(largest)
```

```python
avg_price.plot(kind="bar", color=["tomato", "steelblue"], figsize=(6, 4))
plt.title("Average listing price by preferred area")
plt.xlabel("Preferred area"); plt.ylabel("Average price")
plt.xticks(rotation=0)
plt.tight_layout(); plt.show()
```

# Q8 - Top 5 Products Sold the Most (Pandas)

**Question:** Find the top 5 products that have been sold the most in the past month.

**Dataset:** `Exp8.csv` (columns `product_name`, `quantity`, `order_date`).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose Exp8.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(file_name, encoding="utf-8-sig")
df["order_date"] = pd.to_datetime(df["order_date"], format="mixed", dayfirst=True)

# "Past month" = the latest month present in the data
last = df["order_date"].max().to_period("M")
past_month = df[df["order_date"].dt.to_period("M") == last]
print("Past month:", last, "| rows:", len(past_month))
```

## Step 2: Top 5 products by quantity sold in the past month

```python
top5 = (past_month.groupby("product_name")["quantity"].sum()
        .sort_values(ascending=False).head(5))
print(top5)
```

```python
top5.sort_values().plot(kind="barh", figsize=(9, 4), color="seagreen")
plt.title(f"Top 5 products sold ({last})")
plt.xlabel("Quantity sold"); plt.ylabel("Product")
plt.tight_layout(); plt.show()
```

## Optional: Top 5 products over the whole dataset

```python
print(df.groupby("product_name")["quantity"].sum().sort_values(ascending=False).head(5))
```

# Question 8 – Find the Top 5 Best-Selling Products

**Question:** Use Pandas to find the five products that have been sold in the greatest quantities during the past month.

```python
import pandas as pd

# Sample sales data
sales_data = pd.DataFrame({
    "product_name": ["Pen", "Book", "Pen", "Bag", "Book",
                     "Bottle", "Pencil", "Bag", "Pen", "Bottle"],
    "quantity_sold": [20, 12, 15, 8, 18, 10, 25, 6, 11, 14]
})

# Add quantities for products that appear more than once
top_5_products = (
    sales_data.groupby("product_name")["quantity_sold"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("Top 5 best-selling products:")
print(top_5_products)
```

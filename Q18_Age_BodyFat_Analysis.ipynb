# Q18 - Age and Body Fat Analysis

**Question:** Calculate mean, median and standard deviation of age and %fat using Pandas; draw boxplots; draw a scatter plot and a q-q plot.

**Dataset:** `18bodyfat_FINAL.csv` (columns `Age` and `BodyFat`; BodyFat = %fat).

Run the cells from top to bottom. The first code cell asks you to upload the dataset.

```python
# Step 1: Upload the dataset (a file picker will appear - choose 18bodyfat_FINAL.csv)
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
print("Loaded file:", file_name)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

df = pd.read_csv(file_name)
data = df[["Age", "BodyFat"]]
print(data.shape)
data.head()
```

## Step 2: Mean, median and standard deviation

```python
summary = data.agg(["mean", "median", "std"]).round(2)
summary.columns = ["age", "%fat"]
print(summary)
```

## Step 3: Boxplots

```python
fig, ax = plt.subplots(1, 2, figsize=(9, 5))
ax[0].boxplot(data["Age"]);     ax[0].set_title("Boxplot of Age");   ax[0].set_ylabel("Years")
ax[1].boxplot(data["BodyFat"]); ax[1].set_title("Boxplot of %fat"); ax[1].set_ylabel("% body fat")
plt.tight_layout(); plt.show()
```

## Step 4: Scatter plot

```python
plt.figure(figsize=(7, 5))
plt.scatter(data["Age"], data["BodyFat"], color="steelblue", alpha=0.7)
plt.title("Age vs %fat")
plt.xlabel("Age"); plt.ylabel("%fat")
plt.show()
print("Correlation between age and %fat:", round(data["Age"].corr(data["BodyFat"]), 3))
```

## Step 5: Q-Q plots
First: quantiles of age against quantiles of %fat. Then: each variable against a normal distribution.

```python
# Q-Q plot of age vs %fat (compare matching quantiles)
q = np.linspace(0, 100, 100)
age_q = np.percentile(data["Age"], q)
fat_q = np.percentile(data["BodyFat"], q)

plt.figure(figsize=(6, 6))
plt.scatter(age_q, fat_q, color="steelblue")
lims = [min(age_q.min(), fat_q.min()), max(age_q.max(), fat_q.max())]
plt.plot(lims, lims, "r--", label="y = x")
plt.title("Q-Q plot: Age vs %fat")
plt.xlabel("Age quantiles"); plt.ylabel("%fat quantiles")
plt.legend(); plt.show()

# Normal Q-Q plots
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
stats.probplot(data["Age"], dist="norm", plot=ax[0]);     ax[0].set_title("Normal Q-Q: Age")
stats.probplot(data["BodyFat"], dist="norm", plot=ax[1]); ax[1].set_title("Normal Q-Q: %fat")
plt.tight_layout(); plt.show()
```

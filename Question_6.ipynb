# Question 6 – Calculate Purchase Cost with Discount and Tax

**Question:** A cashier needs to calculate the total cost of a customer's purchase using item prices, quantities, a discount rate, and a tax rate.

```python
# Calculate total purchase cost, including discount and tax

prices = [100, 250, 80]       # Price of each item
quantities = [2, 1, 3]        # Quantity of each item
discount_rate = 10            # Discount percentage
tax_rate = 5                  # Tax percentage

subtotal = sum(price * quantity for price, quantity in zip(prices, quantities))
discount = subtotal * discount_rate / 100
after_discount = subtotal - discount
tax = after_discount * tax_rate / 100
total_cost = after_discount + tax

print("Subtotal: ₹", round(subtotal, 2))
print("Discount: ₹", round(discount, 2))
print("Tax: ₹", round(tax, 2))
print("Total cost: ₹", round(total_cost, 2))
```
